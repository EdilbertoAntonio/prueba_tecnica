```python
import os
if os.path.basename(os.getcwd()) == "limpieza":
    os.chdir("..")
    
import pandas as pd
import numpy as np
from utils.funciones_filtrado import tipo_nulo_unicos_x_columna
```

```python
base_campanas = pd.read_excel('campanas.xlsx')
base_campanas.head()
```

Vamos a estandarizar las columnas de canal y sucursal, ya que encontramos inconsistencias.

```python
base_campanas['canal'] = base_campanas['canal'].str.lower().str.strip() 
base_campanas['sucursal'] = base_campanas['sucursal'].str.lower().str.strip() 

base_campanas['sucursal'] = np.where(base_campanas['sucursal'].str.startswith('a'), 'angelopolis', base_campanas['sucursal'])
```

Para llenar los valores faltantes, supondremos que serán las medias de cada columna, debido a que tenemos pocos datos, y como es una analisis rápido, creo que es la mejor opción.

```python
media_presupuesto = round(base_campanas['presupuesto'].mean(), 2)
media_ventas = round(base_campanas['ventas_generadas'].mean(), 2)

base_campanas['presupuesto'] = base_campanas['presupuesto'].fillna(media_presupuesto)
base_campanas['ventas_generadas'] = base_campanas['ventas_generadas'].fillna(media_ventas)

base_campanas
```

```python
inspeccion = tipo_nulo_unicos_x_columna(base_campanas)
inspeccion
```
