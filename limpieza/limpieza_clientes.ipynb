```python
import os
if os.path.basename(os.getcwd()) == "limpieza":
    os.chdir("..")
    
import pandas as pd
import numpy as np
from utils.funciones_filtrado import tipo_nulo_unicos_x_columna
```

```python
base_clientes = pd.read_excel('clientes.xlsx')
base_clientes
```

Por el primer supuesto vamos a quedarnos con un sólo registro de cada cliente, priorizando el registro donde tenga mayor cantidad de puntos, de este modo no pierde la mayor cantidad de puntos que tiene registrado.
Después convertiremos la columna de fecha al tipo correspondiente, es decir al de fecha y los registros donde exista un número, que sea un valor nulo o sea una entrada de texto, se convertirá a 0

```python
base_clientes = base_clientes.sort_values(by='puntos', ascending=False)
base_clientes = base_clientes[~base_clientes['nombre'].duplicated(keep='first')]
base_clientes['fecha_registro'] = pd.to_datetime(base_clientes['fecha_registro'])
base_clientes['telefono'] = pd.to_numeric(base_clientes['telefono'], errors='coerce').fillna(0).astype('int64')
base_clientes['correo'] = base_clientes['correo'].fillna('S/C')
# aquí podnría una validación, para rectificar que todos los correos sean correos válidos, para evitar que se tengan correos incorrectos
base_clientes
```

Notamos que hay un registro que no tiene nombre, por lo que no valdría quedarnos con ese debido que no nos proporcionó ni un correo ni un número, por lo que se descartaría. 

```python
base_clientes.dropna(subset=['nombre'], inplace=True)
base_clientes
```

```python
no_registros, no_columnas = base_clientes.shape
print(f"Número de productos registrados: {no_registros}, número de columnas de la tabla: {no_columnas}")
```
