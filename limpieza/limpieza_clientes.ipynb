```python
import os
if os.path.basename(os.getcwd()) == "limpieza":
    os.chdir("..")
    
import pandas as pd
import numpy as np
```

```python
base_clientes = pd.read_excel('clientes.xlsx')
base_clientes
```

Por el primer supuesto vamos a quedarnos con un sólo registro de cada cliente, priorizando el registro donde tenga mayor cantidad de puntos.

```python
base_clientes = base_clientes.sort_values(by='puntos', ascending=False)
base_clientes = base_clientes[~base_clientes['nombre'].duplicated(keep='first')]
base_clientes['fecha_registro'] = pd.to_datetime(base_clientes['fecha_registro'])
base_clientes['nombre'] = base_clientes['nombre'].fillna('Sin Nombre')
base_clientes['telefono'] = pd.to_numeric(base_clientes['telefono'], errors='coerce').fillna(0)
base_clientes
```
