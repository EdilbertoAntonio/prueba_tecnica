```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from utils.funciones_filtrado import tipo_dato_x_columna, nulos_x_columna, unicos_x_columna
```

```python
base_productos = pd.read_excel('productos.xlsx')
print(base_productos.head())

no_registros, no_columnas = base_productos.shape
print("\n")
print(f"Número de productos registrados: {no_registros}, número de columnas de la tabla: {no_columnas}")
```

Podemos notar que en la columna 'producto_id' tenemos un id repetido y eso es algo que no debería de suceder debido a que al hacer algún tipo de cruce con otro tabla, vamos a obtener resultados incorrectos, ya que como son dos productos diferentes, asociados a un mismo id, las ventas, stock se asociaran a un sólo producto generando valores más altos y al otro producto tendrá valores bajos o nulos.

De igual manera en la columna 'categoría' observamos que existen diferencias, debido a que en le primer registro se escribió 'café' y en el segundo 'Café', que de igual manera al hacer análisis de esa columna generará errores, ya que, se trarán como diferentes cuando deben ser los mismos.

```python
print(tipo_dato_x_columna(base_productos))
```

Los tipos de datos son consistentes, ya que al menos en esta tabla no esperamos otro tipo de dato, como númerico o de fecha

```python
print('Número de valores nulos por cada columna')
print(nulos_x_columna(base_productos))


print('\n')

print('Número de valores únicos por cada columna')
print(unicos_x_columna(base_productos))
```

Observamos que la columna 'producto_id' aparte de tener un datos repetido, tiene un valor nulo, lo que provoca que la cantidad de valores unicos 16 no coincida con la cantidad de productos registrados en la tabla, que son 18.

También en la columna 'categoria', notamos que sus valores únicos no son 4, como esperamos en la ficha técnica, ya que tenemos 11 valores, es decir, como observamos anteriormente, 'café' y 'Café' los tomará como diferentes, por lo que para otras categorias probablemente se escribió de diferente forma, generando nuevas categorías.

Además en la columna 'activo' notamos que hay 4 valores únicos, cuando yo supondría que sólo fueran 2, verdadero y falso (True y False), posiblemente existan registros mal escritos de la misma forma que sucedió con la columna 'categoría'

```python
#vemos todos los valores unicos de la columna 'categoria'
base_productos['categoria'].unique()
```

Notamos que efectivamente, hay 4 formas en las que se escribió 'Café', 2 formas que se escribió 'Frío', 2 de 'Té' y 3 de 'Alimento', generando las 11 "categorías" falsas

```python
#vemos todos los valores unicos de la columna 'activo'
base_productos['activo'].unique()
```

Notamos que aparte de 'True' y 'False' que ya son los valores booleanos que identifica python, existe el valor 'VERDADEOR' y el valor 'FALSE', por lo que se debería convertir esos valores de a True y False respetivamente

```python
# inspeccionmos la columna de tamaño por si existe alguna inconsistencia como las columnas anteriores
base_productos['tamaño'].unique()
```

Vemos que esta columna no tiene ninguna inconsistencia en cuanto a sus categorias de tamaños, ya que no hay ninguna repetida.
